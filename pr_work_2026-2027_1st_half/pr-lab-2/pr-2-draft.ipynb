# I. Classification Decision Tree

### 1.1. Theoretical Background & Mathematical Foundations
A Decision Tree classifier recursively splits the feature space into axis-aligned rectangular regions. At each internal node, the algorithm seeks the feature index $j$ and split threshold $t$ that maximize the **Information Gain** (or impurity reduction):

$$IG(D, j, t) = I(D) - \left( \frac{|D_L|}{|D|} I(D_L) + \frac{|D_R|}{|D|} I(D_R) \right)$$

Where:
- $D$ is the current set of samples at the node.
- $D_L = \{(x, y) \in D \mid x_j \le t\}$ (left child branch).
- $D_R = \{(x, y) \in D \mid x_j > t\}$ (right child branch).
- $I(\cdot)$ is the impurity measure.

#### Impurity Measures:
1. **Gini Impurity**:
   $$I_{\text{Gini}}(p) = 1 - \sum_{k=1}^{K} p_k^2$$
2. **Entropy**:
   $$I_{\text{Entropy}}(p) = - \sum_{k=1}^{K} p_k \log_2(p_k + \epsilon)$$

Where $p_k = \frac{1}{|D|} \sum_{i \in D} \mathbb{I}(y_i = k)$ is the proportion of class $k$ samples in node $D$, and $\epsilon = 10^{-15}$ avoids $\log_2(0)$.

#### Stopping Criteria:
- Depth reaches `max_depth`.
- Number of samples is less than `min_samples_split`.
- Node is pure ($I(D) = 0$, all samples have the same class label).

# II. Regression Decision Tree

### 2.1. Theoretical Background & Mathematical Foundations
A Regression Decision Tree partitions feature space into regions $R_m$ and predicts a continuous real value $\hat{y}_m \in \mathbb{R}$ for each leaf $m$. The optimal leaf prediction is the arithmetic mean of target values in region $R_m$:

$$\hat{y}_m = \bar{y}_{R_m} = \frac{1}{|D_m|} \sum_{i \in D_m} y_i$$

#### Splitting Criterion (Variance Reduction / Mean Squared Error):
The impurity of a regression node $D$ is measured by its **Variance** (or MSE with respect to its mean):

$$\text{MSE}(D) = \frac{1}{|D|} \sum_{i \in D} (y_i - \bar{y})^2$$

At each node, the optimal split $(j^*, t^*)$ maximizes **Variance Reduction**:

$$\Delta \text{Var}(D, j, t) = \text{MSE}(D) - \left( \frac{|D_L|}{|D|} \text{MSE}(D_L) + \frac{|D_R|}{|D|} \text{MSE}(D_R) \right)$$

#### Stopping Criteria:
- Depth reaches `max_depth`.
- Number of samples is less than `min_samples_split`.
- Variance is near zero (all targets identical).
