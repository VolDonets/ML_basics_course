# I. Implement `Classification` Decision Tree

[YouTube: "Decision and Classification Trees, Clearly Explained"](https://www.youtube.com/watch?v=_L39rN6gz7Y)

```python
from sklearn import datasets

# Fetch the most common classical classification datasets
# Setting return_X_y=True separates the features (X) and labels (y) directly
classification_datasets = {
    "Iris": datasets.load_iris(return_X_y=True),
    "Wine": datasets.load_wine(return_X_y=True),
    "Breast Cancer": datasets.load_breast_cancer(return_X_y=True),
    "Digits": datasets.load_digits(return_X_y=True)
}

# Iterate through them to see their shape
for name, (X, y) in classification_datasets.items():
    print(f"{name} Dataset:")
    print(f" - Features (X) shape: {X.shape}")
    print(f" - Labels (y) shape: {y.shape}\n")
```

# II. Implement `Regression` Decision Tree

[YouTube: "Regression Trees, Clearly Explained"](https://www.youtube.com/watch?v=g9c66TUylZ4)

```python
from sklearn import datasets

# Fetch the most common classical regression datasets
regression_datasets = {
    # 'load_diabetes' and 'load_linnerud' are bundled locally
    "Diabetes": datasets.load_diabetes(return_X_y=True),
    "Linnerud": datasets.load_linnerud(return_X_y=True),

    # 'fetch_california_housing' requires an internet connection the first time it runs
    "California Housing": datasets.fetch_california_housing(return_X_y=True)
}

# Iterate through them to see their shape
for name, (X, y) in regression_datasets.items():
    print(f"{name} Dataset:")
    print(f" - Features (X) shape: {X.shape}")
    print(f" - Target (y) shape: {y.shape}\n")
```
